```python
import pandas as pd
import numpy as np

ratings = pd.read_csv(
    '/content/ratings.csv'
)

links = pd.read_csv(
    '/content/links.csv'
)

movies_catalog = pd.read_csv(
    '/content/top10K-TMDB-movies.csv'
)
```

```python
ratings.head()
```

```python
links.head()
```

```python
movies_catalog.head()
```

```python
print("Ratings shape:", ratings.shape)

print("Unique users:",
      ratings['userId'].nunique())

print("Unique movies:",
      ratings['movieId'].nunique())

print("Missing values:")
print(ratings.isnull().sum())
```

```python
ratings['rating'].value_counts().sort_index()
```

```python
ratings = ratings.merge(
    links[['movieId', 'tmdbId']],
    on='movieId',
    how='inner'
)
```

```python
ratings['tmdbId'] = pd.to_numeric(
    ratings['tmdbId'],
    errors='coerce'
)

ratings = ratings.dropna(
    subset=['tmdbId']
)

ratings['tmdbId'] = ratings['tmdbId'].astype(int)
```

```python
movies_catalog['id'] = pd.to_numeric(
    movies_catalog['id'],
    errors='coerce'
)

movies_catalog = movies_catalog.dropna(
    subset=['id']
)

movies_catalog['id'] = movies_catalog['id'].astype(int)
```

```python
ratings = ratings.merge(
    movies_catalog[['id', 'title']],
    left_on='tmdbId',
    right_on='id',
    how='inner'
)
```

```python
ratings[
    ['userId', 'movieId', 'tmdbId', 'title', 'rating']
].head(10)
```

```python
rating_counts = ratings.groupby(
    'tmdbId'
).size()

eligible_movies = rating_counts[
    rating_counts >= 5
].index

cf_ratings = ratings[
    ratings['tmdbId'].isin(eligible_movies)
].copy()
```

```python
item_user_matrix = cf_ratings.pivot_table(
    index='tmdbId',
    columns='userId',
    values='rating',
    aggfunc='mean'
).fillna(0)
```

```python
item_user_matrix.shape
```

```python
from scipy.sparse import csr_matrix
from sklearn.neighbors import NearestNeighbors
```

```python
item_user_sparse = csr_matrix(
    item_user_matrix.values
)
```

```python
cf_model = NearestNeighbors(
    metric='cosine',
    algorithm='brute',
    n_neighbors=6
)
```

```python
cf_model.fit(item_user_sparse)
```

```python
movie_lookup = (
    cf_ratings[
        ['tmdbId', 'title']
    ]
    .drop_duplicates('tmdbId')
    .set_index('tmdbId')
)
```

```python
def recommend_collaborative(movie_title, n=5):

    matches = movie_lookup[
        movie_lookup['title'].str.lower()
        == movie_title.lower()
    ]

    if matches.empty:
        return "Movie not found in rating dataset."

    movie_id = matches.index[0]

    movie_index = item_user_matrix.index.get_loc(
        movie_id
    )

    distances, indices = cf_model.kneighbors(
        item_user_sparse[movie_index],
        n_neighbors=min(n + 1, len(item_user_matrix))
    )

    recommendations = []

    for idx, distance in zip(
        indices.flatten(),
        distances.flatten()
    ):

        recommended_id = item_user_matrix.index[idx]

        if recommended_id == movie_id:
            continue

        title = movie_lookup.loc[
            recommended_id, 'title'
        ]

        similarity_score = 1 - distance

        recommendations.append({
            'Movie': title,
            'Similarity': round(similarity_score, 3)
        })

        if len(recommendations) == n:
            break

    return pd.DataFrame(recommendations)
```

```python
recommend_collaborative('The Godfather')
```

```python
recommend_collaborative('Iron Man')
```

```python
import pickle

collaborative_objects = {
    'model': cf_model,
    'matrix': item_user_matrix,
    'sparse_matrix': item_user_sparse,
    'movie_lookup': movie_lookup
}

with open(
    '/content/collaborative_model.pkl',
    'wb'
) as file:
    pickle.dump(collaborative_objects, file)

print("Collaborative model saved successfully!")
```

```python
import os
import zipfile

# Project files to include
project_files = [
    "README.md",
    "app.py",
    "app_backup.py",
    "app_integrated.py",
    "collaborative_model.pkl",
    "datasets.zip",
    "links.csv",
    "movies.csv",
    "movies_list.pkl",
    "ratings.csv",
    "similarity.pkl",
    "tags.csv",
    "top10K-TMDB-movies.csv"
]

# Create backup folder
backup_folder = "/content/Movie_Recommendation_System_Backup"
os.makedirs(backup_folder, exist_ok=True)

zip_path = "/content/Movie_Recommendation_System_Complete.zip"

print("Preparing project backup...\n")

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zip_file:

    for filename in project_files:

        source_path = os.path.join("/content", filename)

        if os.path.isfile(source_path):
            zip_file.write(
                source_path,
                arcname=filename
            )
            size_mb = os.path.getsize(source_path) / (1024 * 1024)
            print(f"Added: {filename} ({size_mb:.2f} MB)")

        else:
            print(f"Not found: {filename}")

print("\nBackup ZIP created successfully!")
print("Location:", zip_path)

size_mb = os.path.getsize(zip_path) / (1024 * 1024)
print(f"ZIP size: {size_mb:.2f} MB")
```
